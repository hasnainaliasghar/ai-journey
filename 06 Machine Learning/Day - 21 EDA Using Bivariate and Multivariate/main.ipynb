## 1. Scattor Plot Numerical - Numerical combo

```python
sns.scatterplot(data = tips, x = "total_bill", y = "tip", hue = "sex", style="smoker", size = "size")
```

## 2. BarPlot Numerical to Categorial

```python
sns.barplot(data=titanic, x = "Pclass", y = "Age", hue="Sex")
```

# Box Plot -> Numerical to Cate

```python
sns.boxplot(data = titanic , x = titanic["Sex"], y="Age")
```

```python
sns.displot(data=titanic[titanic["Survived"] == 0], x="Age")
```

```python
sns.heatmap(pd.crosstab(titanic["Pclass"], titanic["Survived"]),
            annot=True,
            fmt="d")
```

```python
(titanic.groupby("Sex")["Survived"].mean() * 100).plot(kind = "bar")
```

```python
sns.clustermap(pd.crosstab(titanic["Parch"], titanic["Survived"]))
```
