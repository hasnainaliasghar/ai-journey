```python
plt.figure(figsize=(14,4))
plt.subplot(121)
sns.histplot(data=df, x = "cgpa", kde = True)
plt.subplot(122)
sns.histplot(data=df, x = "placement_exam_marks", kde = True)
```

```python
Upper_limit = float(df["cgpa"].mean() + 3*df["cgpa"].std())
Lower_Limit = float(df["cgpa"].mean() - 3*df["cgpa"].std())
float(Upper_limit), float(Lower_Limit)
```

```python
df[(df["cgpa"] > 8.80) | (df["cgpa"] < 5.11)]
```

```python
df["cgpa"] = np.where(
    df["cgpa"] > Upper_limit,
    Upper_limit,
    np.where(
        df["cgpa"] < Lower_Limit,
        Lower_Limit,
        df["cgpa"]
    )
)
```
