```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.model_selection import train_test_split

import importlib
import SGDRegressor as sgdr_module
importlib.reload(sgdr_module)

from SGDRegressor import SGDRegressor

import numpy as np
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
reg = LinearRegression()
reg.fit(X_train, y_train)
reg.coef_, reg.intercept_
```

```python
y_pred = reg.predict(X_test)
r2_score(y_test, y_pred)
```

```python
sgd = SGDRegressor(leraning_rate=0.01, epochs=100)
sgd.fit(X_train, y_train)
```

```python
y_pred = sgd.predict(X_test)
```

```python
r2_score(y_test, y_pred)
```

```python
for lr_val in [0.001, 0.01, 0.05, 0.1,0.026]:
    sgd2 = SGDRegressor(leraning_rate=lr_val, epochs=300)
    sgd2.fit(X_train, y_train)
    y_pred = sgd2.predict(X_test)
    print(f"lr={lr_val}: R2 = {r2_score(y_test, y_pred)}")
```
