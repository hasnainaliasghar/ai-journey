## **Example 3**
Nurel Network

CGPA | Placement

without autograd

```python
import torch
```

```python
#Cgpa against placement
x = torch.tensor(6.7)
y = torch.tensor(0.0)

#Random values/ intital value of bias against weight
w = torch.tensor(1.0)
b = torch.tensor(0.0)
```

```python
#Binary Loss Calculate
def binary_cross_entropy_loss(prediction, target):
    epsilon = 1e-8
    prediction = torch.clamp(prediction, epsilon, 1 - epsilon)
    loss = -(target * torch.log(prediction) + (1 - target) * torch.log(1 - prediction))
    return loss.mean()
```

```python
#forwar propogation
z = w * x + b
y_pred = torch.sigmoid(z)
loss = binary_cross_entropy_loss(y_pred, y)
```

```python
loss
```

```python
#backpropagation
dloss_dy_pred =(y_pred - y)/(y_pred * (1-y_pred))

dy_pred_dz = y_pred*(1 - y_pred)

dz_dw = x
dz_db = 1

dl_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dl_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
print(f"Manual gradient of loss w.r.t weight {dl_dw}")
print(f"Manual gradient of loss w.r.t bias {dl_db}")
```

```python
#clearing grad
x = torch.tensor(2.0, requires_grad=True)
y = (x**2).mean()
y.backward()
print(x.grad)
```
