```python
import torch
print(torch.__version__)
print(torch.cuda.is_available())
```

# Creating Tensors

```python
# using Empty
a = torch.empty(2,3)
```

```python
#check type
type(a)
```

```python
# using ones
torch.ones(2,3)
```

```python
torch.zeros(2,3)
```

```python
torch.rand(2,3)
```

```python
torch.manual_seed(100)
torch.rand(2,3)
```

```python
torch.tensor([[1,2,3],[4,5,6]])
```

```python
#other ways

#arange
torch.arange(0,18,2).reshape(3,3)

#Linspace
torch.linspace(0,18,12).reshape(3,4)

#Logspace
torch.logspace(-9,10,15)
```

# Tensors Shape

```python
x = torch.tensor([[1,2,3],[4,5,6]])
x.shape
```

```python
torch.empty_like(x)
```

```python
torch.zeros_like(x)
```

```python
torch.ones_like(x)
```

```python
torch.rand_like(x,dtype=torch.float64)
```
