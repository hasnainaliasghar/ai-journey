# Training PipeLine on Breast Cancer DataSet


## Basic Imports

```python
import pandas as pd
import numpy as np

import torch
import torch.nn as nn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.drop(columns=["id","Unnamed: 32"],inplace=True)
df.head()
```

```python
df.shape
```

## train_test_split

```python
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=["diagnosis"]),df["diagnosis"],test_size=0.2)
```

```python
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
X_train
```

```python
le = LabelEncoder()
y_train = le.fit_transform(y_train)
y_test = le.transform(y_test)
```

```python
y_train
```

## Numpy to Tensors

```python
# Converting train/test array to tensors
X_train_tensor = torch.from_numpy(X_train.astype(np.float32))
X_test_tensor = torch.from_numpy(X_test.astype(np.float32))
y_train_tensor = torch.from_numpy(y_train.astype(np.float32))
y_test_tensor = torch.from_numpy(y_test.astype(np.float32))
```

```python
X_test_tensor.shape
```

```python
from torch.utils.data import Dataset, DataLoader
class CustomDataset(Dataset):
  def __init__(self, x_tensor, y_tensor):
    self.x = x_tensor
    self.y = y_tensor

  def __getitem__(self, index):
    return (self.x[index], self.y[index])

  def __len__(self):
    return len(self.x)
```

```python
train_dataset = DataLoader(CustomDataset(X_train_tensor, y_train_tensor), batch_size=32, shuffle=True)
test_dataset = DataLoader(CustomDataset(X_test_tensor, y_test_tensor), batch_size=32, shuffle=False)
```

## Defining the Model

```python
class NurelNetwork(nn.Module):
  def __init__(self,num_features):
    super().__init__()
    self.linear = nn.Linear(num_features,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)
    return out

  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

## Important Parameters

```python
learning_rate = 0.1
epochs = 25
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'Using device: {device}')
```

## Training Pipeline

```python
Loss_Function = nn.BCELoss()
```

```python
model = NurelNetwork(X_train_tensor.shape[1])

#optim
optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

#forward Pass
for epoch in range(epochs):
  for batch_feature, batch_label in train_dataset:
    y_pred = model(batch_feature)

    #Loss Calculation
    loss = Loss_Function(y_pred,batch_label.view(-1,1))

    #zero grad
    optimizer.zero_grad()

    #BackwardPass
    loss.backward()

    #parameters update
    optimizer.step()
  print(f'Epoch: {epoch+1}, Loss: {loss.item()}')
```

# Evaluation

```python
# Model evaluation using test_loader
model.eval()  # Set the model to evaluation mode
accuracy_list = []

with torch.no_grad():
    for batch_features, batch_labels in test_dataset:
        # Forward pass
        y_pred = model(batch_features)
        y_pred = (y_pred > 0.8).float()  # Convert probabilities to binary predictions

        # Calculate accuracy for the current batch
        batch_accuracy = (y_pred.view(-1) == batch_labels).float().mean().item()
        accuracy_list.append(batch_accuracy)

# Calculate overall accuracy
overall_accuracy = sum(accuracy_list) / len(accuracy_list)
print(f'Accuracy: {overall_accuracy:.4f}')
```
